a) $\begin{cases}x+y=3,\\ 2x+2y=6.\end{cases}$   $$\begin{bmatrix} 1 & 1 & \vert & 3 \\ 2 & 2 & \vert & 6 \end{bmatrix} \xrightarrow{L_2 \leftarrow L_2 - 2L_1} \begin{bmatrix} 1 & 1 & \vert & 3 \\ 0 & 0 & \vert & 0 \end{bmatrix}$$
Como obteve-se 0x+0y=0 é um sistema possível e indeterinado com infinitas soluções.

b) $\begin{cases}x+y=3,\\ 2x+2y=7.\end{cases}$   $$\begin{bmatrix} 1 & 1 & \vert & 3 \\ 2 & 2 & \vert & 7 \end{bmatrix} \xrightarrow{L_2 \leftarrow L_2 - 2L_1} \begin{bmatrix} 1 & 1 & \vert & 3 \\ 0 & 0 & \vert & 1 \end{bmatrix}$$
Como obteve-se 0x+0y=1, ou seja, 0=1 é um sistema impossível, sem solução.

c) $\begin{cases}x+y=3,\\ 2x-y=0.\end{cases}$   $$\begin{bmatrix} 1 & 1 & \vert & 3 \\ 2 & -1 & \vert & 0 \end{bmatrix} \xrightarrow{L_2 \leftarrow L_2 - 2L_1} \begin{bmatrix} 1 & 1 & \vert & 3 \\ 0 & -3 & \vert & -6 \end{bmatrix}$$
Logo, o sistema é possível e determinado com o resultado (x,y)=(1,2).

```python
import numpy as np

def eliminacao_gauss(A, b):
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)

    # 1. Formar a matriz aumentada [A|b]
    Ab = np.hstack([A, b.reshape(-1, 1)])

    # 2. Eliminação direta (eliminar elementos abaixo dos pivôs)
    for i in range(n):
        pivo = Ab[i, i]
        if pivo == 0:
            raise ValueError(f"Pivô nulo encontrado na posição ({i},{i}).")

        for j in range(i + 1, n):
            m = Ab[j, i] / pivo
            Ab[j, i:] -= m * Ab[i, i:]

    # 3. Substituição regressiva
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        soma = np.dot(Ab[i, i+1:n], x[i+1:n])
        x[i] = (Ab[i, -1] - soma) / Ab[i, i]

    return x, Ab

# Sistema do Exercício 2
A2 = [[2, 1, -1],
      [-3, -1, 2],
      [-2, 1, 2]]

b2 = [8, -11, -3]

# Resolução com o algoritmo personalizado
sol_algoritmo, Ab_escalonada = eliminacao_gauss(A2, b2)

# Resolução com numpy.linalg.solve
sol_numpy = np.linalg.solve(A2, b2)

print("Matriz Aumentada Escalonada:")
print(Ab_escalonada)
print("\nSolução calculada (Algoritmo):", sol_algoritmo)
print("Solução calculada (NumPy):    ", sol_numpy)
```

A solução calculada pelo procedimento numérico coincide exatamente com a resolução manual e com a resposta dada pela função numpy.linalg.solve, confirmando a precisão do método.

```python
import numpy as np

# Matriz e vetor do sistema de balanço térmico
A_termico = np.array([[4, -1, 0],
                      [-1, 4, -1],
                      [0, -1, 3]], dtype=float)

b_termico = np.array([15, 10, 10], dtype=float)

# Utilizando o algoritmo de Eliminação de Gauss implementado
sol_T, _ = eliminacao_gauss(A_termico, b_termico)

print(f"Temperaturas calculadas: T1 = {sol_T[0]:.2f}, T2 = {sol_T[1]:.2f}, T3 = {sol_T[2]:.2f}")
```
