```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
a1=2.; b1=1.; c1=-1.; d1=8.
a2=-3.; b2=-1.; c2=2.; d2=-11.
a3=-2.; b3=1.; c3=2.; d3=-3.
```

```python
plt.rcParams["animation.html"] = "jshtml"

# Sistema inicial [a, b, c, d]
A = np.array([[a1, b1, c1, d1],
              [a2, b2, c2, d2],
               [a3, b3, c3, d3]])
CORES = ['#4C78A8', '#F58518', '#54A24B']

# Eliminação de Gauss (gerar etapas)
etapas = [A.copy()]
M = A.copy()
for i in range(3):
    for j in range(i + 1, 3):
        if M[j, i] != 0:
            M[j] -= (M[j, i] / M[i, i]) * M[i]
            etapas.append(M.copy())

# Configuração do gráfico
fig = plt.figure(figsize=(10, 4))
ax_mat = fig.add_subplot(121)
ax_3d = fig.add_subplot(122, projection='3d')

def desenhar_plano(ax, eq, cor):
    a, b, c, d = eq
    u, v = np.meshgrid(np.linspace(-8, 8, 10), np.linspace(-8, 8, 10))
    dom = np.argmax(np.abs([a, b, c]))
    if dom == 0 and a: ax.plot_surface((d - b*u - c*v)/a, u, v, color=cor, alpha=0.3)
    elif dom == 1 and b: ax.plot_surface(u, (d - a*u - c*v)/b, v, color=cor, alpha=0.3)
    elif c: ax.plot_surface(u, v, (d - a*u - b*v)/c, color=cor, alpha=0.3)

def atualizar(frame):
    ax_mat.clear(); ax_3d.clear()

    # Tabela com matriz
    ax_mat.axis('off')
    ax_mat.table(cellText=np.round(etapas[frame], 2), colLabels=['x','y','z','b'], loc='center')
    ax_mat.set_title(f"Passo {frame + 1}")

    # Planos 3D
    for k in range(3):
        desenhar_plano(ax_3d, etapas[frame][k], CORES[k])
    ax_3d.set_xlim(-8, 8); ax_3d.set_ylim(-8, 8); ax_3d.set_zlim(-8, 8)

anim = FuncAnimation(fig, atualizar, frames=len(etapas), interval=1200)
plt.close(fig)
HTML(anim.to_jshtml())
anim.save("eliminacao_gauss.gif", writer=PillowWriter(fps=1))
plt.close(fig)
```
