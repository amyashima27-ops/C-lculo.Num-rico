# Tarefa investigativa: erros em métodos de interpolação

### Aplicação: reconstrução da temperatura em uma barra aquecida

Uma barra homogênea de comprimento $L=6$ m está sujeita a temperaturas prescritas nas extremidades e recebe calor de uma fonte distribuída em seu interior. Sensores medem a temperatura em posições discretas. A equipe precisa estimar a temperatura em posições onde não há sensor e avaliar a incerteza dessas estimativas.

Esta atividade combina um modelo simplificado de transferência de calor com interpolação numérica. O objetivo não é resolver a equação diferencial de calor por diferenças finitas. O perfil analítico será usado como referência, para que os erros de diferentes métodos de interpolação possam ser medidos e comparados.

Você vai comparar **polinômio global de Lagrange** e uma **spline cúbica**, primeiro com dados exatos e depois com medições perturbadas por ruído.


## 1. Problema físico e equação de transferência de calor

### 1.1 Modelo físico

Considere uma barra sólida homogênea, com comprimento $L=6$ m e seção transversal constante. A temperatura varia principalmente ao longo do eixo $x$. As extremidades são mantidas a temperaturas conhecidas. No interior, uma fonte térmica volumétrica distribui energia ao longo da barra.

A equação geral de energia térmica, para propriedades constantes e sem mudança de fase, pode ser escrita como

$$\rho c_p\frac{\partial T}{\partial t}=k\nabla^2T+\dot q^{\prime\prime\prime}-\rho c_p\,\mathbf{u}\cdot\nabla T,$$

em que $T$ é a temperatura (K ou °C para diferenças), $t$ é o tempo (s), $\rho$ é a massa específica (kg m$^{-3}$), $c_p$ é o calor específico (J kg$^{-1}$ K$^{-1}$), $k$ é a condutividade térmica (W m$^{-1}$ K$^{-1}$), $\dot q^{\prime\prime\prime}$ é a geração volumétrica de calor (W m$^{-3}$) e $\mathbf{u}$ representa o movimento de um fluido, quando há advecção.

Para esta atividade, adotamos uma **barra sólida em regime estacionário**, sem escoamento interno, com condução unidimensional e geração de calor. Assim, $\partial T/\partial t=0$, $\mathbf{u}=0$ e $\nabla^2T=d^2T/dx^2$. A equação fica

$$k\frac{d^2T}{dx^2}+\dot q^{\prime\prime\prime}(x)=0,\qquad 0<x<L,$$

com condições de contorno de Dirichlet

$$T(0)=18\ ^\circ\mathrm C,\qquad T(L)=20.4\ ^\circ\mathrm C.$$

O primeiro termo representa a condução: diferenças espaciais de temperatura geram fluxo de calor. O termo $\dot q^{\prime\prime\prime}(x)$ representa energia adicionada por unidade de volume e de tempo. O sinal positivo de $\dot q^{\prime\prime\prime}$ indica geração interna de calor.

### 1.2 Perfil de referência usado no experimento

Para criar um caso com solução conhecida, definimos $\xi=x/L$ e adotamos

$$T(x)=18+2.4\,\xi+2.5\sin(\pi\xi),\qquad \xi=\frac{x}{L},\quad 0\le x\le L.$$

A parcela linear garante as temperaturas prescritas nas extremidades. A parcela senoidal representa um aquecimento distribuído mais intenso no interior e nulo nas bordas. Derivando duas vezes em relação à coordenada física $x$,

$$\frac{d^2T}{dx^2}=-\frac{2.5\pi^2}{L^2}\sin\left(\frac{\pi x}{L}\right).$$

Portanto, esse perfil satisfaz a equação estacionária se a geração volumétrica for

$$\dot q^{\prime\prime\prime}(x)=\frac{2.5\,k\pi^2}{L^2}\sin\left(\frac{\pi x}{L}\right).$$

Como $\sin(\pi x/L)\ge0$ dentro da barra, a fonte é não negativa. Não precisamos escolher um valor numérico para $k$: o objetivo é verificar a coerência do perfil e estudar a interpolação. A função analítica é uma referência idealizada; em uma aplicação experimental, seria preciso estimar propriedades, fontes, perdas e condições de contorno.

### 1.3 Questão investigativa

**Como a resposta depende da quantidade e posição dos sensores e do ruído de medição?**



### O que significa erro de interpolação?

Em um ponto de avaliação $x$, o erro assinado é $e(x)=p(x)-T(x)$. O erro absoluto é $|e(x)|$. Como a interpolação reproduz os nós, o erro nesses pontos pode ser praticamente zero; por isso, avaliaremos também muitos pontos **entre** os nós.

Usaremos: erro máximo absoluto $E_{max}=\max_j|e_j|$; erro médio absoluto (MAE) $=\frac{1}{m}\sum_j|e_j|$; raiz do erro quadrático médio (RMSE) $=\sqrt{\frac{1}{m}\sum_j e_j^2}$. O RMSE penaliza mais erros grandes.

Spline quadrática

```python
def spline_quadratica_coef(x, y):
    """Spline quadrática C1 com condição c_0=0; retorna a,b,c locais."""
    x = np.asarray(x, float); y = np.asarray(y, float)
    n = len(x)-1
    if n < 1 or len(y) != len(x) or np.any(np.diff(x) <= 0):
        raise ValueError("x deve ser crescente e ter o mesmo tamanho de y.")
    A = np.zeros((3*n, 3*n)); rhs = np.zeros(3*n); row = 0
    # S_i(x_i)=y_i
    for i in range(n):
        A[row, i] = 1; rhs[row] = y[i]; row += 1
    # S_i(x_(i+1))=y_(i+1)
    for i in range(n):
        h = x[i+1]-x[i]
        A[row, i] = 1; A[row, n+i] = h; A[row, 2*n+i] = h*h
        rhs[row] = y[i+1]; row += 1
    # Continuidade da primeira derivada nos nós internos
    for i in range(n-1):
        h = x[i+1]-x[i]
        A[row, n+i] = 1; A[row, 2*n+i] = 2*h; A[row, n+i+1] = -1
        row += 1
    # Condição de contorno: c_0=0
    A[row, 2*n] = 1
    coef = np.linalg.solve(A, rhs)
    return coef[:n], coef[n:2*n], coef[2*n:]

def spline_quadratica_avalia(x, coef, xq):
    a, b, c = coef
    xq = np.asarray(xq, dtype=float)
    idx = np.clip(np.searchsorted(x, xq, side="right")-1, 0, len(a)-1)
    t = xq-x[idx]
    return a[idx] + b[idx]*t + c[idx]*t**2

def criar_metodos(x, y):
    """Retorna dicionário: interpoladores callable para quatro métodos."""
    lagrange = BarycentricInterpolator(x, y)
    linear = interp1d(x, y, kind="linear", bounds_error=True)
    quad_coef = spline_quadratica_coef(x, y)
    quad = lambda z: spline_quadratica_avalia(x, quad_coef, z)
    cubica = CubicSpline(x, y, bc_type="natural")
    return {"Lagrange global": lagrange,
            "Spline linear": linear,
            "Spline quadrática": quad,
            "Spline cúbica natural": cubica}

def calcular_metricas(metodos, x_teste, y_verdadeiro):
    linhas = []
    for nome, metodo in metodos.items():
        erro = np.asarray(metodo(x_teste))-np.asarray(y_verdadeiro)
        linhas.append({"Método": nome,
                       "Erro máximo (°C)": np.max(np.abs(erro)),
                       "MAE (°C)": np.mean(np.abs(erro)),
                       "RMSE (°C)": np.sqrt(np.mean(erro**2))})
    return pd.DataFrame(linhas).sort_values("RMSE (°C)").reset_index(drop=True)

print("Funções de interpolação e métricas definidas.")
```

## 3. Métodos de interpolação e métricas de erro

O interpolador global de Lagrange passa por todos os sensores usando um único polinômio. Usamos a forma baricêntrica, que evita expandir diretamente os coeficientes e costuma ser numericamente mais estável.

As splines constroem polinômios por intervalos. A spline linear conecta pontos por retas. A quadrática deste notebook impõe continuidade da função e da primeira derivada e usa a condição de contorno $c_0=0$ no primeiro intervalo. A spline cúbica natural impõe continuidade até a segunda derivada e segunda derivada nula nas extremidades.

Para quantificar a diferença, calculamos $e(x)=T_{interp}(x)-T_{ref}(x)$. O erro máximo mostra o pior desvio na malha. O MAE dá o desvio absoluto médio. O RMSE penaliza mais os erros de grande magnitude. Todos têm unidade de graus Celsius.

A malha fina serve como conjunto de pontos de avaliação: os valores nela não são novos sensores. Assim, a comparação mede desempenho entre os nós.

```python
def spline_quadratica_coef(x, y):
    """Spline quadrática C1 com condição c_0=0; retorna a,b,c locais."""
    x = np.asarray(x, float); y = np.asarray(y, float)
    n = len(x)-1
    if n < 1 or len(y) != len(x) or np.any(np.diff(x) <= 0):
        raise ValueError("x deve ser crescente e ter o mesmo tamanho de y.")
    A = np.zeros((3*n, 3*n)); rhs = np.zeros(3*n); row = 0
    # S_i(x_i)=y_i
    for i in range(n):
        A[row, i] = 1; rhs[row] = y[i]; row += 1
    # S_i(x_(i+1))=y_(i+1)
    for i in range(n):
        h = x[i+1]-x[i]
        A[row, i] = 1; A[row, n+i] = h; A[row, 2*n+i] = h*h
        rhs[row] = y[i+1]; row += 1
    # Continuidade da primeira derivada nos nós internos
    for i in range(n-1):
        h = x[i+1]-x[i]
        A[row, n+i] = 1; A[row, 2*n+i] = 2*h; A[row, n+i+1] = -1
        row += 1
    # Condição de contorno: c_0=0
    A[row, 2*n] = 1
    coef = np.linalg.solve(A, rhs)
    return coef[:n], coef[n:2*n], coef[2*n:]

def spline_quadratica_avalia(x, coef, xq):
    a, b, c = coef
    xq = np.asarray(xq, dtype=float)
    idx = np.clip(np.searchsorted(x, xq, side="right")-1, 0, len(a)-1)
    t = xq-x[idx]
    return a[idx] + b[idx]*t + c[idx]*t**2

def criar_metodos(x, y):
    """Retorna dicionário: interpoladores callable para quatro métodos."""
    lagrange = BarycentricInterpolator(x, y)
    linear = interp1d(x, y, kind="linear", bounds_error=True)
    quad_coef = spline_quadratica_coef(x, y)
    quad = lambda z: spline_quadratica_avalia(x, quad_coef, z)
    cubica = CubicSpline(x, y, bc_type="natural")
    return {"Lagrange global": lagrange,
            "Spline linear": linear,
            "Spline quadrática": quad,
            "Spline cúbica natural": cubica}

def calcular_metricas(metodos, x_teste, y_verdadeiro):
    linhas = []
    for nome, metodo in metodos.items():
        erro = np.asarray(metodo(x_teste))-np.asarray(y_verdadeiro)
        linhas.append({"Método": nome,
                       "Erro máximo (°C)": np.max(np.abs(erro)),
                       "MAE (°C)": np.mean(np.abs(erro)),
                       "RMSE (°C)": np.sqrt(np.mean(erro**2))})
    return pd.DataFrame(linhas).sort_values("RMSE (°C)").reset_index(drop=True)

print("Funções de interpolação e métricas definidas.")
```

## Etapa 2 — Número e posição dos sensores

Nesta etapa, investigue somente o polinômio global de Lagrange. A interpolação polinomial global pode oscilar entre nós, sobretudo perto das extremidades, quando se usam muitos nós equiespaçados. A distribuição de Chebyshev concentra nós nas bordas e pode reduzir esse efeito.

Para $N$ nós de Chebyshev-Lobatto em $[a,b]$, use

$$x_k=\frac{a+b}{2}+\frac{b-a}{2}\cos\left(\frac{k\pi}{N-1}\right),\quad k=0,\ldots,N-1.$$

Ordene as posições crescentemente. Compare $N=5,9,13,17$ para nós equiespaçados e de Chebyshev. Para cada caso, avalie o erro em todos os pontos da malha fina e registre o erro máximo e o RMSE.

**Atenção:** uma redução de erro neste perfil particular não prova que o mesmo ocorrerá para qualquer função. Relacione a observação à suavidade e à estrutura do perfil usado.

```python
def nos_chebyshev_lobatto(a, b, N):
    k = np.arange(N)
    z = np.cos(k*np.pi/(N-1))
    return np.sort((a+b)/2 + (b-a)*z/2)

resultados_etapa2 = []
Ns = [5, 9, 13, 17]

for N in Ns:
    # 1. Nós equiespaçados
    x_eq = np.linspace(x_min, x_max, N)
    y_eq = temperatura_real(x_eq)
    lag_eq = BarycentricInterpolator(x_eq, y_eq)
    erro_eq = lag_eq(x_fino) - T_fino

    # 2. Nós de Chebyshev-Lobatto
    x_cheb = nos_chebyshev_lobatto(x_min, x_max, N)
    y_cheb = temperatura_real(x_cheb)
    lag_cheb = BarycentricInterpolator(x_cheb, y_cheb)
    erro_cheb = lag_cheb(x_fino) - T_fino

    resultados_etapa2.append({
        "N": N,
        "E_max Equiespaçado (°C)": np.max(np.abs(erro_eq)),
        "RMSE Equiespaçado (°C)": np.sqrt(np.mean(erro_eq**2)),
        "E_max Chebyshev (°C)": np.max(np.abs(erro_cheb)),
        "RMSE Chebyshev (°C)": np.sqrt(np.mean(erro_cheb**2))
    })

tabela_nos = pd.DataFrame(resultados_etapa2)
display(tabela_nos)
```

A função de temperatura \(T(x) = 18 + 2.4ξ + 2.5\sin(pi*ξ)\) é contínua e suave, o que permite que os dois tipos de nós (equiespaçados e de Chebyshev) apresentem bons resultados na interpolação. Entretanto, os nós de Chebyshev proporcionam maior estabilidade numérica e reduzem os erros nas extremidades do intervalo, evitando o fenômeno de Runge. Assim, conseguem atingir uma precisão próxima à precisão de máquina (\(10^{-15}\,°C\)) utilizando menos pontos do que os nós equiespaçados

```python
import time

N_valores = np.arange(4, 41)
dados_etapa3 = []

for N in N_valores:
    # Nós
    x_eq = np.linspace(x_min, x_max, N)
    y_eq = temperatura_real(x_eq)
    x_cheb = nos_chebyshev_lobatto(x_min, x_max, N)
    y_cheb = temperatura_real(x_cheb)

    # 1. Lagrange Equiespaçado
    t0 = time.perf_counter()
    m_eq = BarycentricInterpolator(x_eq, y_eq)
    t_const_eq = time.perf_counter() - t0

    t0 = time.perf_counter()
    pred_eq = m_eq(x_fino)
    t_aval_eq = time.perf_counter() - t0
    err_eq = pred_eq - T_fino

    # 2. Lagrange Chebyshev
    t0 = time.perf_counter()
    m_cheb = BarycentricInterpolator(x_cheb, y_cheb)
    t_const_cheb = time.perf_counter() - t0

    t0 = time.perf_counter()
    pred_cheb = m_cheb(x_fino)
    t_aval_cheb = time.perf_counter() - t0
    err_cheb = pred_cheb - T_fino

    # 3. Spline Cúbica Natural
    t0 = time.perf_counter()
    m_cub = CubicSpline(x_eq, y_eq, bc_type="natural")
    t_const_cub = time.perf_counter() - t0

    t0 = time.perf_counter()
    pred_cub = m_cub(x_fino)
    t_aval_cub = time.perf_counter() - t0
    err_cub = pred_cub - T_fino

    dados_etapa3.append({
        "N": N,
        "Emax_Eq": np.max(np.abs(err_eq)), "RMSE_Eq": np.sqrt(np.mean(err_eq**2)),
        "Emax_Cheb": np.max(np.abs(err_cheb)), "RMSE_Cheb": np.sqrt(np.mean(err_cheb**2)),
        "Emax_Cub": np.max(np.abs(err_cub)), "RMSE_Cub": np.sqrt(np.mean(err_cub**2)),
        "T_total_Eq": t_const_eq + t_aval_eq,
        "T_total_Cheb": t_const_cheb + t_aval_cheb,
        "T_total_Cub": t_const_cub + t_aval_cub
    })

df_e3 = pd.DataFrame(dados_etapa3)

# Gráficos em escala logarítmica
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5))

# Erro Máximo
axes[0].semilogy(df_e3["N"], df_e3["Emax_Eq"], 'o-', markersize=4, label="Lagrange Equiesp.")
axes[0].semilogy(df_e3["N"], df_e3["Emax_Cheb"], 's-', markersize=4, label="Lagrange Chebyshev")
axes[0].semilogy(df_e3["N"], df_e3["Emax_Cub"], '^-', markersize=4, label="Spline Cúbica")
axes[0].set_title("Erro Máximo ($E_{max}$) vs $N$")
axes[0].set_xlabel("Número de sensores ($N$)")
axes[0].set_ylabel("Erro Máximo (°C) [log]")
axes[0].grid(True, which="both", alpha=0.3)
axes[0].legend()

# RMSE
axes[1].semilogy(df_e3["N"], df_e3["RMSE_Eq"], 'o-', markersize=4, label="Lagrange Equiesp.")
axes[1].semilogy(df_e3["N"], df_e3["RMSE_Cheb"], 's-', markersize=4, label="Lagrange Chebyshev")
axes[1].semilogy(df_e3["N"], df_e3["RMSE_Cub"], '^-', markersize=4, label="Spline Cúbica")
axes[1].set_title("RMSE vs $N$")
axes[1].set_xlabel("Número de sensores ($N$)")
axes[1].set_ylabel("RMSE (°C) [log]")
axes[1].grid(True, which="both", alpha=0.3)

# Tempo de Execução
axes[2].plot(df_e3["N"], df_e3["T_total_Eq"]*1e3, 'o-', markersize=4, label="Lagrange Equiesp.")
axes[2].plot(df_e3["N"], df_e3["T_total_Cheb"]*1e3, 's-', markersize=4, label="Lagrange Chebyshev")
axes[2].plot(df_e3["N"], df_e3["T_total_Cub"]*1e3, '^-', markersize=4, label="Spline Cúbica")
axes[2].set_title("Tempo Total (Construção + Avaliação)")
axes[2].set_xlabel("Número de sensores ($N$)")
axes[2].set_ylabel("Tempo (ms)")
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

O cálculo dos pesos baricêntricos do polinômio de Lagrange possui custo \(O(N^2)\), enquanto cada avaliação depende da quantidade de pontos da malha (\(M = 2001\)). Já a spline cúbica tem custo \(O(N)\) para sua construção, pois resolve um sistema tridiagonal, sendo uma opção mais eficiente computacionalmente quando o número de pontos aumenta.
Não existe um método que seja melhor em todas as situações, pois o desempenho depende da quantidade e da distribuição dos pontos.
- Para \(N\) pequeno a moderado (\(N \leq 15\)), o Lagrange com nós de Chebyshev apresenta uma convergência mais rápida, alcançando uma precisão próxima à precisão de máquina (\(\approx 10^{-15}\,°C\)) com poucos pontos. A spline cúbica também apresenta bons resultados, mas sua convergência é mais lenta.
- Para \(N\) grande (\(N > 25\)), o Lagrange com nós equiespaçados começa a apresentar problemas de estabilidade numérica, principalmente nas extremidades do intervalo, onde os erros de arredondamento podem ser amplificados. Já a spline cúbica e o Lagrange com nós de Chebyshev mantêm maior estabilidade, apresentando resultados mais confiáveis nessa situação.

## **Recomendação Técnica: Reconstrução do Perfil de Temperatura na Barra**

1\. Escolha do método com poucos sensores:

 Quando temos poucos sensores (\\(N \approx 5\\) a \\(9\\)) e os dados não apresentam ruído, o método de **Lagrange** é uma boa escolha, principalmente com nós de Chebyshev-Lobatto, caso seja possível escolher a posição dos sensores. Se os sensores já estiverem distribuídos de maneira uniforme, também é possível utilizar nós equiespaçados. Como o perfil de temperatura apresenta uma variação suave, o polinômio consegue representar bem seu comportamento, com erros menores quando comparado às splines lineares e cúbicas.

2\. O que acontece quando aumentamos o número de sensores:

Quando temos muitos sensores, o polinômio de Lagrange com nós equiespaçados pode começar a apresentar oscilações, principalmente nas extremidades da barra. Isso acontece devido ao fenômeno de Runge e à amplificação dos erros numéricos. Nesse caso, a **Spline Cúbica** passa a ser uma opção mais interessante, pois apresenta maior estabilidade e consegue representar o perfil sem gerar oscilações tão acentuadas.

3\. Quando existem erros nas medições:

 Na prática, os sensores podem apresentar pequenas variações e erros de medição. Nessa situação, utilizar métodos de interpolação exata pode não ser a melhor escolha, pois eles fazem a curva passar por todos os pontos, inclusive aqueles que possuem ruído. Isso pode prejudicar o resultado e gerar oscilações que não representam o comportamento real da temperatura. Por isso, é mais adequado utilizar métodos de aproximação, como mínimos quadrados ou splines de suavização, que reduzem a influência desses erros.

4\. Cuidados nas extremidades da barra (**\\(x=0\\)**&#x20;e&#x20;**\\(x=L\\)**):

- Posicionamento dos sensores:  É importante instalar sensores nas duas extremidades da barra para conhecer as temperaturas nos limites do intervalo e evitar a necessidade de extrapolação. Também é interessante ter mais pontos próximos às bordas, já que essas regiões podem apresentar maiores erros de interpolação.
- Condição de contorno da spline:  A spline cúbica natural considera que a segunda derivada nas extremidades é igual a zero. Para o perfil estudado, essa condição é adequada, mas em situações reais isso pode não acontecer. Por isso, é importante escolher a condição de contorno de acordo com o comportamento físico do problema.

5\. Limitações e diferença entre interpolação e extrapolação:

- Limitações em situações reais: Quando não conhecemos a função exata da temperatura, não conseguimos calcular diretamente o erro máximo (\\(E\_{max}\\)) e o erro quadrático médio (RMSE). Nesse caso, podemos avaliar a confiabilidade do método retirando alguns pontos e verificando se ele consegue estimar seus valores corretamente. Também é necessário considerar os possíveis erros dos sensores e variações de temperatura que podem não ter sido registradas.
- Interpolação e extrapolação: Os resultados deste estudo são válidos para a interpolação, ou seja, para estimar temperaturas dentro do intervalo em que temos sensores. Já a extrapolação consiste em estimar temperaturas fora desse intervalo e exige mais cuidado, pois os métodos podem apresentar erros maiores. Os polinômios podem apresentar grandes oscilações e as splines podem não representar corretamente o comportamento da temperatura fora da região analisada.
